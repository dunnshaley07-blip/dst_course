# Working with Images Using Pillow

In this notebook, we explore image operations such as cropping, resizing, pasting, and transforming images.

## Import Library

```python
from PIL import Image
```

## Load Image

```python
img = Image.open("baboon.png")
img
```

## Image Format Information

We can inspect the file format and a more detailed description.

```python
img.format
```

```python
img.format_description
```

```python
img.size
```

## Cropping an Image

Crop a rectangular region from the image.

```python
cropped = img.crop((0, 0, 200, 200))
cropped
```

Access individual pixels using `.load()`

```python
pixels = cropped.load()
pixels[20,20]
```

```python
tinycrop = img.crop((0, 0, 5, 5))
big = tinycrop.resize((img.width, img.height), resample=Image.NEAREST)
big.show()
```

## Copying an Image

Create a copy of the image.

```python
img_copy = img.copy()
img_copy
```

## Resizing an Image

Resize the image to a smaller size.

```python
small = img.resize((100, 100))
small
```

## Pasting One Image Onto Another

Paste the small image onto the original.

```python
img_paste = img.copy()
img_paste.paste(small, (0, 0))
img_paste
```

## Creating a Wallpaper Effect

Tile the small image across a larger canvas.

```python
wallpaper = Image.new('RGB', img.size)
w, h = small.size
for y in range(0, img.size[1], h):
    for x in range(0, img.size[0], w):
        wallpaper.paste(small, (x, y))
wallpaper
```

## Rotating an Image

```python
rotated = img.rotate(45)
rotated
```

## Flipping an Image

Flip the image vertically and horizontally.

```python
flipped_lr = img.transpose(Image.FLIP_LEFT_RIGHT)
flipped_lr
```

```python
flipped_tb = img.transpose(Image.FLIP_TOP_BOTTOM)
flipped_tb
```

## Transpose Operations

Other transpose options include rotations by 90-degree increments.

```python
rot90 = img.transpose(Image.ROTATE_90)
rot90
```

```python
rot180 = img.transpose(Image.ROTATE_180)
rot180
```

```python
rot270 = img.transpose(Image.ROTATE_270)
rot270
```

## Save an Output Image

```python
wallpaper.save("baboon_wallpaper.png")
```
